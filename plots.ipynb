```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df_bl = pd.read_csv('2_SMOTE_Boderline_data.csv')

expected_cols = [
    "Type",
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
    "Failure Type",
]
missing = [c for c in expected_cols if c not in df_bl.columns]
counts_ft = df_bl["Failure Type"].value_counts()
type_counts = df_bl["Type"].value_counts()

# 1) Distribution of Failure Types
plt.figure(figsize=(8,6))
plt.bar(counts_ft.index.astype(str), counts_ft.values)
plt.title("Distribution of Failure Types (SMOTE-Borderline)")
plt.xticks(rotation=45)
plt.xlabel("Failure Type")
plt.ylabel("Count")
plt.show()

# 2) Air Temperature Distribution
plt.figure(figsize=(8,6))
plt.hist(df_bl["Air temperature [K]"].dropna().values, bins=30)
plt.title("Air Temperature Distribution (SMOTE-Borderline)")
plt.xlabel("Air Temperature [K]")
plt.ylabel("Count")
plt.show()

# 3) Process Temperature Distribution
plt.figure(figsize=(8,6))
plt.hist(df_bl["Process temperature [K]"].dropna().values, bins=30)
plt.title("Process Temperature Distribution (SMOTE-Borderline)")
plt.xlabel("Process Temperature [K]")
plt.ylabel("Count")
plt.show()

# 4) Rotational Speed Distribution
plt.figure(figsize=(8,6))
plt.hist(df_bl["Rotational speed [rpm]"].dropna().values, bins=40)
plt.title("Rotational Speed Distribution (SMOTE-Borderline)")
plt.xlabel("Rotational Speed [rpm]")
plt.ylabel("Count")
plt.show()

# 5) Torque Distribution
plt.figure(figsize=(8,6))
plt.hist(df_bl["Torque [Nm]"].dropna().values, bins=40)
plt.title("Torque Distribution (SMOTE-Borderline)")
plt.xlabel("Torque [Nm]")
plt.ylabel("Count")
plt.show()

# 6) Tool Wear Distribution
plt.figure(figsize=(8,6))
plt.hist(df_bl["Tool wear [min]"].dropna().values, bins=30)
plt.title("Tool Wear Distribution (SMOTE-Borderline)")
plt.xlabel("Tool Wear [min]")
plt.ylabel("Count")
plt.show()

# 7) Correlation Heatmap (numeric columns only)
num_df = df_bl.select_dtypes(include=[np.number])
corr = num_df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
im = plt.imshow(corr, aspect='auto')
plt.colorbar(im, fraction=0.046, pad=0.04)
plt.xticks(ticks=range(len(corr.columns)), labels=corr.columns, rotation=45, ha='right')
plt.yticks(ticks=range(len(corr.columns)), labels=corr.columns)
plt.title("Correlation Heatmap (SMOTE-Borderline)")
plt.show()

# 8) Torque by Failure Type (Boxplot)
groups = [df_bl.loc[df_bl["Failure Type"] == ft, "Torque [Nm]"].dropna().values
          for ft in counts_ft.index]
plt.figure(figsize=(8,6))
plt.boxplot(groups, labels=counts_ft.index.astype(str), showfliers=False)
plt.title("Torque by Failure Type (SMOTE-Borderline)")
plt.xlabel("Failure Type")
plt.ylabel("Torque [Nm]")
plt.xticks(rotation=45)
plt.show()

# 9) Rotational Speed vs Torque by Failure Type (Scatter)
plt.figure(figsize=(8,6))
for ft in counts_ft.index:
    subset = df_bl[df_bl["Failure Type"] == ft]
    plt.scatter(subset["Rotational speed [rpm]"], subset["Torque [Nm]"], alpha=0.6, label=str(ft))
plt.title("Rotational Speed vs Torque by Failure Type (SMOTE-Borderline)")
plt.xlabel("Rotational Speed [rpm]")
plt.ylabel("Torque [Nm]")
plt.legend(title="Failure Type", loc="best")
plt.show()

# 10) Air vs Process Temperature by Failure Type (Scatter)
plt.figure(figsize=(8,6))
for ft in counts_ft.index:
    subset = df_bl[df_bl["Failure Type"] == ft]
    plt.scatter(subset["Air temperature [K]"], subset["Process temperature [K]"], alpha=0.6, label=str(ft))
plt.title("Air vs Process Temperature by Failure Type (SMOTE-Borderline)")
plt.xlabel("Air Temperature [K]")
plt.ylabel("Process Temperature [K]")
plt.legend(title="Failure Type", loc="best")
plt.show()

# 11) Pie chart of Type column
plt.figure(figsize=(6,6))
plt.pie(type_counts.values, labels=type_counts.index.astype(str), autopct="%1.1f%%", startangle=90)
plt.title("Distribution of Equipment Types (SMOTE-Borderline)")
plt.show()

```
